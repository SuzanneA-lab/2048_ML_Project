```python
#used some template code from https://iproyal.com/blog/scrape-text-from-website/
#used some template code from https://medium.com/free-code-camp/better-web-scraping-in-python-with-selenium-beautiful-soup-and-pandas-d6390592e251
#to run, run pip install beautifulsoup4 and pip install requests prior to execution

from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
import re
import pandas as pd
import os

import random
import time

url = "https://game-difficulty.github.io/2048EndgameTablebase/"

# create a new Firefox session
driver = webdriver.Firefox()
driver.implicitly_wait(30)
driver.get(url)
num_moves=1
num_games=1
dataframelist = []

max_turns = 100
max_games = 10

while num_games<=max_games:
    time.sleep(3)
    print("Game", num_games)

    num_moves = 1
    while num_moves<=max_turns: 
        boardinputbefore = driver.find_element(By.ID, 'board-input')
        boardinputvalbf = boardinputbefore.get_attribute('value')

        onestepbutton = driver.find_element(By.ID, 'btn-step')
        onestepbutton.click()
        time.sleep(0.05)

        boardinput = driver.find_element(By.ID, 'board-input')
        boardinputval = boardinput.get_attribute('value')

        soup = BeautifulSoup(driver.page_source, "html.parser")
        #board_state = soup.find("input",id="board-input")
        score = soup.find("div",id="score")
        #board_state = 

        #print(boardinputval)
        #print(score.get_text())
        #print(num_moves)

        #Define all board states in required format 

        A1 = int(boardinputval[0], 16)
        A2 = int(boardinputval[1], 16)
        A3 = int(boardinputval[2], 16)
        A4 = int(boardinputval[3], 16)

        B1 = int(boardinputval[4], 16)
        B2 = int(boardinputval[5], 16)
        B3 = int(boardinputval[6], 16)
        B4 = int(boardinputval[7], 16)

        C1 = int(boardinputval[8], 16)
        C2 = int(boardinputval[9], 16)
        C3 = int(boardinputval[10], 16)
        C4 = int(boardinputval[11], 16)

        D1 = int(boardinputval[12], 16)
        D2 = int(boardinputval[13], 16)
        D3 = int(boardinputval[14], 16)
        D4 = int(boardinputval[15], 16)

        g_bf = [int(c, 16) for c in boardinputvalbf]
        g_af = [int(c, 16) for c in boardinputval]

        move="0"
        if g_bf != g_af:
            # Sum index positions of active tiles (idx // 4 is row, idx % 4 is col)
            row_shift = sum(i // 4 for i, val in enumerate(g_af) if val > 0) - sum(i // 4 for i, val in enumerate(g_bf) if val > 0)
            col_shift = sum(i % 4 for i, val in enumerate(g_af) if val > 0) - sum(i % 4 for i, val in enumerate(g_bf) if val > 0)
            
            if abs(row_shift) > abs(col_shift):
                move = "DOWN" if row_shift > 0 else "UP"
            else:
                move = "RIGHT" if col_shift > 0 else "LEFT"
    
        Movedata = {
            "TileA1": A1,
            "TileA2": A2,
            "TileA3": A3,
            "TileA4": A4,
            "TileB1": B1,
            "TileB2": B2,
            "TileB3": B3,
            "TileB4": B4,
            "TileC1": C1,
            "TileC2": C2,
            "TileC3": C3,
            "TileC4": C4,
            "TileD1": D1,
            "TileD2": D2,
            "TileD3": D3,
            "TileD4": D4,
            "Move": move,
            "Game_Number": num_games,
            "Game_Turn": num_moves,
            "Game_Score": score.get_text()
        }

        dataframelist.append(Movedata)

        num_moves += 1

    
    num_games += 1

    #generate random locations for 1s
    first2 = random.randint(0,15)
    second2 = random.randint(0,15)
    startstate = ""

    for i in range(0, 16):
        if i!=first2 and i!=second2:
            startstate += "0" 
        else:
            startstate += "1"
    
    boardinput.clear()
    boardinput.send_keys(startstate)
    setboardbutton = driver.find_element(By.ID, 'btn-set')
    setboardbutton.click()

appended_data = pd.DataFrame(dataframelist)
appended_data.to_excel("..\data\data.xlsx")

print(appended_data)
driver.quit()

#setboardinput = driver.find_element(By.ID, 'board-input')
#setboardinput.clear()
#setboardbutton = driver.find_element(By.ID, 'btn-set')

#setboardinput.send_keys("0000000000000011");
#setboardbutton.click() #click fhsu link

#import requests 
#from bs4 import BeautifulSoup

# 1. Send an HTTP GET request
#response = requests.get(url)

# 2. Parse the HTML content
#soup = BeautifulSoup(response.text, "html.parser")

# 3. Locate target elements
#quotes = soup.find_all("div", class_="score-box")
#authors = soup.find_all("small", class_="author")

# 4. Extract and clean the text
#for i in range(len(quotes)):
#    print(f"Score: {quotes[i].text}")
    #print(f"AUTHOR: {authors[i].text}")
#    print("-" * 20)
```
